```python
from batch import *

# BV and Ru(bpy)3 stock solutions now made in pH 6 buffer
# mixing limitations in cuvette, colour gradient seen

# V(Ru(bpy)3, 1 mM, buffer) = 200 ul,
# V(EDTA, saturated, buffer) = 1 ml,
# made up to 3 ml pH 6 buffer
# blue 455 nm irradiation

x = 'B'
duration = 15*60 # 15 min
cstock = 1e-3 # [] in stock solution
vsol = 3e-3 # total volume of reaction mixture
root = f"./results/260701_Rubpy_{x}V_EDTA_Buffered/Rubpy_{x}V_"

# (V(EV, 1 mM, buffer) in ul used, start times) for each run, start time adjusted slightly to begin at linear region
runs = [(300, "11-46-40-000"), # 11-46-26-000 recorded start time
        (600, "12-21-16-000"), # 12-21-16-000
        (900, "12-55-19-000"), # 12-55-19-000
        (1200, "13-36-05-000"), # 13-36-05-000
        (1500, "14-12-42-000"), # 14-12-42-000
        (1800, "16-40-13-000")] # 16-40-13-000

for (v, start) in runs:
    data = f'{root}{v}_EDTA_Buffered'
    c0 = solprop(cstock=cstock, valiquot=v*1e-6, vsol=vsol) # initial concentration
    concvstimeplot(data, c0, start, x, duration=duration,
                   istart=100, r2threshold=0.99, interceptthreshold=1e-6)

for (v, start) in runs:
    data = f'{root}{v}_EDTA_Buffered'
    c0 = solprop(cstock=cstock, valiquot=v*1e-6, vsol=vsol) # initial concentration
    lncvstimeplot(data, c0, start, x, duration=duration,
                  istart=100, r2threshold=0.99, interceptthreshold=np.log(c0))

for (v, start) in runs:
    data = f'{root}{v}_EDTA_Buffered'
    c0 = solprop(cstock=cstock, valiquot=v*1e-6, vsol=vsol) # initial concentration
    spectrumvstimeplot(data, c0, start, x, duration=duration, vmin=0, vmax=1)

spectrumvsconcplot(root, runs, x, t=5*60)
```
