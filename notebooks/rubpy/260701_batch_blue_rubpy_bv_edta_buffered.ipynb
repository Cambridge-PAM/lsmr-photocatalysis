```python
from lsmr_photocatalysis import Batch, X
import numpy as np

"""
Batch reaction in the lightbox with blue 455 nm irradiation.
BV2+ and [Ru(bpy)3]2+ stock solutions now made in pH 6 buffer.
Mixing limitations in cuvette, colour gradient seen.

V([Ru(bpy)3]2+, 1 mM, pH 6 buffer) = 200 ul (67 uM),
V(BV2+, 1 mM, pH 6 buffer) = 300-1800 ul (100-600 uM),
V(EDTA, saturated <25 mM, pH 6 buffer) = 1 ml (<8333 uM),
Made up to 3 ml with pH 6 buffer.
"""

x = 'BVrubpybatch'
duration = 15*60 # plot up to 15 min
t = 5*60 # see spectrum at 5 min

# (V(BV2+, 1 mM, pH 6 buffer) in ul used, start time, istart) for each run, istart adjusted to begin regression at the linear region
runs = [(300, '11-46-26-000', 30),
        (600, '12-21-16-000', 0),
        (900, '12-55-19-000', 0),
        (1200, '13-36-05-000', 0),
        (1500, '14-12-42-000', 0),
        (1800, '16-40-13-000', 0)]

expts = []
for (v, start, istart) in runs:
    data = f'260701_Rubpy_BV_EDTA_Buffered/Rubpy_BV_{v}_EDTA_Buffered' # data folder name
    c0 = 1e-3*(v*1e-6)/3e-3 # initial BV2+ concentration (1 mM stock) x (v ul aliquot) / (3 ml cuvette)
    details = f'{X['RU']} = 67 μM\n[EDTA] $\\leq$ 8333 μM' # concentrations of reagents
    expts.append((Batch(data, start, c0, details, x, correction=True, duration=duration), istart))

for expt, istart in expts:
    expt.concvstimeplot(trial=100, r2threshold=0.998, interceptthreshold=1e-6, istart=istart)
for expt, istart in expts:
    expt.concvstimeplot(ln=True, trial=100, r2threshold=0.998, interceptthreshold=np.log(expt.c0), istart=istart)
for expt, _ in expts:
    expt.spectrumvstimeplot(amin=0, amax=1)
Batch.spectrumvsconcplot([expt for expt, _ in expts], t, inspect=3)
```
