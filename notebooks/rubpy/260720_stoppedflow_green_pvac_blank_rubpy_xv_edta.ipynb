```python
from stopflow import *

# stop flow with reaction in 50 ul chip and blank PVAc film
# green 565 nm irradiation at 11.4 mW for each time point, then uv-vis immediately after in the same chip
# flushed with fresh starting reaction mixture and repeated for new time point

# V(Ru(bpy)3, 12.5 mM, buffer) = 240 ul
# V(XV, 25 mM, buffer) = 720 ul
# V(EDTA, saturated, buffer) = 540 ul

t = [t*30 for t in range(1, 15)] # residence times in seconds for 1-7 minutes in 30s intervals
c0 = solprop(cstock=25e-3, valiquot=720e-6, vsol=1.5e-3) # initial XV2+ concentration
film = 'blank PVAc'
light = 'green'
wavelength = 565e-9 # m
power = 11.4e-3 # W

for x, i in [('E', None), ('B', None)]:
    data = f'./results/260720_Rubpy_{x}V_EDTA_Blank_Green_StopFlow/'
    concvstimeplot(data, t, c0, x, film, light, wavelength, power, i=i)
    lncvstimeplot(data, t, c0, x, film, light, wavelength, power, i=i)
    spectrumvstimeplot(data, t, c0, x, film, light, wavelength, power)
```
