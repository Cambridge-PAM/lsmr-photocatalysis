```python
from lsmr_photocatalysis import Batch, LIGHTBOXUV

"""
Batch reaction in the lightbox with UV 340 nm irradiation.
UV degradation test on [Ru(bpy)3]2+.

V(Ru(bpy)3, 1 mM, pH 6 buffer) = 200 ul (67 uM),
Made up to 3 ml pH 6 buffer.
"""

data = "260722_Rubpy_UV"
start = '11-44-00-000' # 11-44-00-000 recorded start time

c0 = 1e-3*200e-6/3e-3 # initial [Ru(bpy)3]2+ concentration (1 mM stock) x (200 ul aliquot) / (3 ml cuvette)
details = 'UV degradation test'

x = 'Rubpy'
light = 'UV'
wavelength = 340e-9 # m
power = LIGHTBOXUV # W

duration = 30*60 # plot up to 30 min
t = 5*60 # see spectrum at 5 min

expt = Batch(data, start, c0, details, x, light=light, wavelength=wavelength, power=power,
             correction=True, duration=duration)
expt.concvstimeplot()
expt.spectrumvstimeplot(amin=0, amax=1.1)
Batch.spectrumvsconcplot([expt], t)
```
