```python
from lsmr_photocatalysis import FIGSIZE, DPI, X
import matplotlib.pyplot as plt
import numpy as np

conditions = ['Green, Blank', 'Green, Coumarin 153', 'Blue, Blank', 'Blue, Coumarin 153']
colors = ['lightgreen', 'seagreen', 'skyblue', 'steelblue']

ev = [0.0078, 0.0075, 0.018, 0.020]
everr = [0.0006, 0.0004, 0.003, 0.001]
evpos = np.arange(0, len(conditions))

fig, ax = plt.subplots(figsize=FIGSIZE, dpi=DPI)

evbar = ax.bar(evpos, ev, yerr=everr, capsize=2, color=colors, label=conditions)
ax.bar_label(evbar, label_type='center', fmt='%.4f')

ax.spines['top'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.tick_params(axis='y', length=0)
ax.grid(axis='y', alpha=0.35)
ax.set_axisbelow(True)

ax.set_xticks([])
ax.set_xticklabels([])
ax.set_ylabel('$\\Phi$ / %')
ax.set_title('Apparent quantum yields $\\Phi$ / % \n to test the LDS performance of Coumarin 153 PVAc films')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.05), ncol=2, frameon=False)
plt.show()
```
