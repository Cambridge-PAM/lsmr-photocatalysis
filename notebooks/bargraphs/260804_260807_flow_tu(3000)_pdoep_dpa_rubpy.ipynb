```python
from lsmr_photocatalysis import FIGSIZE, DPI, X
import matplotlib.pyplot as plt
import numpy as np

viologens = ['EV', 'BV']
conditions = ['Blue, Blank', 'Blue, PdOEP/DPA', 'Green, Blank', 'Green, PdOEP/DPA']
colors = ['skyblue', 'steelblue', 'lightgreen', 'seagreen']

ev = [0.27, 0.26, 0.13, 0.12]
everr = [0.03, 0.03, 0.02, 0.02]
evpos = np.arange(0, len(conditions))

bv = [0.74, 0.71, 0.30, 0.27]
bverr = [0.06, 0.05, 0.03, 0.03]
bvpos = np.arange(len(conditions), len(conditions)*2)+1

fig, ax = plt.subplots(figsize=FIGSIZE, dpi=DPI)

evbar = ax.bar(evpos, ev, yerr=everr, capsize=2, color=colors, label=conditions)
bvbar = ax.bar(bvpos, bv, yerr=bverr, capsize=2, color=colors)

ax.bar_label(evbar, label_type='center', fmt='%.2f')
ax.bar_label(bvbar, label_type='center', fmt='%.2f')

ax.spines['top'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.tick_params(axis='y', length=0)
ax.grid(axis='y', alpha=0.35)
ax.set_axisbelow(True)

ax.set_xticks([np.mean(evpos), np.mean(bvpos)])
ax.set_xticklabels(viologens)
ax.set_ylabel('$\\Phi$ / %')
ax.set_title('Apparent quantum yields $\\Phi$ / % \n to test the TTA-UC performance of PdOEP/DPA TU(3000) films')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=2, frameon=False)
plt.show()
```
