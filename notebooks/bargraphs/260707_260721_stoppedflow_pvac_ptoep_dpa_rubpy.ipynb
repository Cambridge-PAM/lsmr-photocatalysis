```python
from lsmr_photocatalysis import FIGSIZE, DPI, X
import matplotlib.pyplot as plt
import numpy as np

viologens = ['EV', 'BV']
conditions = ['Blue, No film', 'Blue, Blank', 'Blue, PtOEP/DPA', 'Green, Blank', 'Green, PtOEP/DPA']
colors = ['slategrey', 'skyblue', 'steelblue', 'lightgreen', 'seagreen']

ev = [0.175, 0.17, 0.134, 0.052, 0.042]
everr = [0.007, 0.01, 0.005, 0.003, 0.002]
evpos = np.arange(0, len(conditions))

bv = [0.47, 0.43, 0.32, 0.097, 0.081]
bverr = [0.02, 0.02, 0.01, 0.004, 0.002]
bvpos = np.arange(len(conditions), len(conditions)*2)+1

fig, ax = plt.subplots(figsize=FIGSIZE, dpi=DPI)

evbar = ax.bar(evpos, ev, yerr=everr, capsize=2, color=colors, label=conditions)
bvbar = ax.bar(bvpos, bv, yerr=bverr, capsize=2, color=colors)

ax.bar_label(evbar, label_type='center', fmt='%.2f')
ax.bar_label(bvbar, label_type='center', fmt='%.2f')

ax.spines['top'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.tick_params(axis='y', length=0)
ax.grid(axis='y', alpha=0.35)
ax.set_axisbelow(True)

ax.set_xticks([np.mean(evpos), np.mean(bvpos)])
ax.set_xticklabels(viologens)
ax.set_ylabel('$\\Phi$ / %')
ax.set_title('Apparent quantum yields $\\Phi$ / % \n to test the TTA-UC performance of PtOEP/DPA PVAc films')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=2, frameon=False)
plt.show()
```
