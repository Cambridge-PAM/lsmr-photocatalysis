```python
from lsmr_photocatalysis.settings import FIGSIZE, DPI, X
import matplotlib.pyplot as plt
import numpy as np

WIDTH = 0.35
runs = np.arange(100, 601, 100)

ev = [0.109, 0.174, 0.224, 0.456, 0.504, 0.553]
# everr = [0.003, 0.002, 0.004, 0.004, 0.006, 0.002] # error bars too small to be plotted

bv = [0.225, 0.392, 0.431, 0.480, 0.736, 0.884]
# bverr = [0.002, 0.003, 0.003, 0.002, 0.002, 0.003] # error bars too small to be plotted

fig, ax = plt.subplots(figsize=FIGSIZE, dpi=DPI)

ax.bar(np.arange(0, len(runs))-WIDTH*1.2/2, ev, width=WIDTH, label='EV', color='lightsteelblue')
ax.bar(np.arange(0, len(runs))+WIDTH*1.2/2, bv, width=WIDTH, label='BV', color='slategray')

ax.spines['top'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.tick_params(axis='y', length=0)
ax.grid(axis='y', alpha=0.35)
ax.set_axisbelow(True)

ax.set_xticks(list(range(len(runs))))
ax.set_xticklabels(runs)
ax.set_xlabel(f'{X['XV0']} / μM')
ax.set_ylabel('$\\Phi$ / %')
ax.set_title(f'Apparent quantum yield $\\Phi$ / % against {X['XV0']} / μM\nscreening with {X['RU']} = 67 μM, [EDTA] = 8333 μM')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, frameon=False)
plt.show()
```
