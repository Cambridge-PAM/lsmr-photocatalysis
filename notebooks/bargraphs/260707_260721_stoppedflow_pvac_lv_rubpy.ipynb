```python
from lsmr_photocatalysis import FIGSIZE, DPI, X
import matplotlib.pyplot as plt
import numpy as np

viologens = ['EV', 'BV']
conditions = ['Blue, No film', 'Blue, Blank', 'Blue, Lumogen Violet', 'UV, Blank', 'UV, Lumogen Violet']
colors = ['slategrey', 'skyblue', 'steelblue', 'mediumpurple', 'darkorchid']

ev = [0.175, 0.17, 0.144, 0.26, 0.20]
everr = [0.007, 0.01, 0.007, 0.01, 0.01]
evpos = np.arange(0, len(conditions))

bv = [0.47, 0.43, 0.37, 0.36, 0.295]
bverr = [0.02, 0.02, 0.01, 0.01, 0.007]
bvpos = np.arange(len(conditions), len(conditions)*2)+1

fig, ax = plt.subplots(figsize=FIGSIZE, dpi=DPI)

evbar = ax.bar(evpos, ev, yerr=everr, capsize=2, color=colors, label=conditions)
bvbar = ax.bar(bvpos, bv, yerr=bverr, capsize=2, color=colors)

ax.bar_label(evbar, label_type='center', fmt='%.2f')
ax.bar_label(bvbar, label_type='center', fmt='%.2f')

ax.spines['top'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.tick_params(axis='y', length=0)
ax.grid(axis='y', alpha=0.35)
ax.set_axisbelow(True)

ax.set_xticks([np.mean(evpos), np.mean(bvpos)])
ax.set_xticklabels(viologens)
ax.set_ylabel('$\\Phi$ / %')
ax.set_title('Apparent quantum yields $\\Phi$ / % \n to test the LDS performance of Lumogen Violet PVAc films')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=2, frameon=False)
plt.show()
```
